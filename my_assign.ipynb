```python
import nltk
corpus="""Hello,I am Rashmitha's Nakirikanti.
How are you?.I am too good!"""
```

```python
from nltk.tokenize import sent_tokenize
from nltk.tokenize import word_tokenize
from nltk.tokenize import wordpunct_tokenize
from nltk.tokenize import TreebankWordTokenizer
```

```python
sent=sent_tokenize(corpus)
print(f"sent_tokenize:{sent}")
word=word_tokenize(corpus)
print(f"word_tokenize:{word}")
punc_word=wordpunct_tokenize(corpus)
print(f"wordpunc_tokenize:{punc_word}")
tree=TreebankWordTokenizer()
fi=tree.tokenize(corpus)
print(f"Treetokenizer:{fi}")
```
